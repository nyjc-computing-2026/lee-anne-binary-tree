```python
# Task 1.1
class BSTNode:
    def __init__(self, isbn, title): # Initialising
        self.left_child = None
        self.right_child = None # left and right child is None for empty BST
        self.isbn = isbn
        self.title = title

    def add(self, isbn, title):
        """Adds a book record beginnning frmo current node"""
        if isbn == self.isbn: # if ISBN already exists
            print('A book with this ISBN number already exists.')
            
        elif self.left_child is None and self.right_child is None: # If it is a leaf node
            if isbn < self.isbn:
                self.left_child = BSTNode(isbn, title)
            elif isbn > self.isbn:
                self.right_child = BSTNode(isbn, title)
                
        elif self.left_child is None: # If no left child and key < current, key will be the new left child
            if isbn < self.isbn:
                self.left_child = BSTNode(isbn, title)
            else:
                self.right_child.add(isbn, title)
                
        elif self.right_child is None: # If no left child and key > current, key will be the new left child
            if isbn > self.isbn:
                self.right_child = BSTNode(isbn, title)
            else:
                self.left_child.add(isbn, title)

        else:
            if isbn < self.isbn:
                self.left_child.add(isbn, title) # Traversing recursively
            else:
                self.right_child.add(isbn, title)
                
    def find(isbn):
        """Searches the subtree, returns None if no matching record found"""
        if self.isbn == isbn:
            return self.title
        elif isbn < self.isbn: # Following the left child
            self = self.left_child
        else:
            self = self.right_child # Following the right child
    
```
